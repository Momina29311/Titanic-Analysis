```python
import numpy as np
import pandas as pd

df = pd.read_csv("data/titanic.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df.describe(include="all")
```

```python
overall_survival_rate = df["Survived"].mean() * 100
print(f"Overall survival rate: {overall_survival_rate:.2f}%")

df["Survived"].value_counts()
```

```python
gender_survival = df.groupby("Sex")["Survived"].agg(
    passengers="count",
    survivors="sum",
    survival_rate="mean"
)

gender_survival["survival_rate"] = gender_survival["survival_rate"] * 100
gender_survival.sort_values("survival_rate", ascending=False)
```

```python
class_survival = df.groupby("Pclass")["Survived"].agg(
    passengers="count",
    survivors="sum",
    survival_rate="mean"
)

class_survival["survival_rate"] = class_survival["survival_rate"] * 100
class_survival.sort_values("survival_rate", ascending=False)
```

```python
first_class_women = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

print("First-class women:", len(first_class_women))
print("Survival rate:", first_class_women["Survived"].mean() * 100)

first_class_women[["Name", "Age", "Fare", "Survived"]].head(10)
```

```python
third_class_men = df[(df["Sex"] == "male") & (df["Pclass"] == 3)]

print("Third-class men:", len(third_class_men))
print("Survival rate:", third_class_men["Survived"].mean() * 100)

third_class_men[["Name", "Age", "Fare", "Survived"]].head(10)
```

```python
top_fares = df.sort_values(by="Fare", ascending=False)

top_fares[["Name", "Pclass", "Sex", "Age", "Fare", "Survived"]].head(10)
```

```python
print(df["Age"].agg(["mean", "median", "min", "max", "std"]))

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teen", "Young Adult", "Adult", "Senior"]
)

age_survival = df.groupby("Age_Group", observed=True)["Survived"].agg(
    passengers="count",
    survival_rate="mean"
)

age_survival["survival_rate"] = age_survival["survival_rate"] * 100
age_survival
```

```python
df["Family_Size"] = df["SibSp"] + df["Parch"] + 1

family_survival = df.groupby("Family_Size")["Survived"].agg(
    passengers="count",
    survival_rate="mean"
)

family_survival["survival_rate"] = family_survival["survival_rate"] * 100
family_survival.sort_values("survival_rate", ascending=False)
```

```python
embarked_survival = df.groupby("Embarked")["Survived"].agg(
    passengers="count",
    survival_rate="mean"
)

embarked_survival["survival_rate"] = embarked_survival["survival_rate"] * 100
embarked_survival.sort_values("survival_rate", ascending=False)
```

```python
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare", "Family_Size"]

df[numeric_cols].corr()["Survived"].sort_values(ascending=False)
```
